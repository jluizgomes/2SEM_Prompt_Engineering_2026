### Slide 06 — BaseModel — definindo um schema em Python

````python
!pip install langchain-ollama langchain-core langchain-classic -q

from langchain_ollama import ChatOllama
from google.colab import userdata
import os

# Definir a API key via variável de ambiente (Colab Secrets)
os.environ["OLLAMA_HOST"]    = "https://ollama.com"
os.environ["OLLAMA_API_KEY"] = userdata.get("OLLAMA_API_KEY")
````

````python
!pip install pydantic -q  # já vem com langchain, mas explicitando

from pydantic import BaseModel, Field, ValidationError
from typing import List, Optional

# Schema: cada campo tem tipo e descrição
class Curriculo(BaseModel):
    nome:         str               = Field(description="Nome completo do candidato")
    email:        str               = Field(description="E-mail do candidato")
    habilidades:  List[str]         = Field(description="Lista de habilidades técnicas")
    anos_exp:     int               = Field(description="Anos de experiência profissional")
    senioridade:  Optional[str]    = Field(None, description="junior/pleno/senior")

# Instanciar — Pydantic valida os tipos automaticamente
curriculo = Curriculo(
    nome="Ana Silva",
    email="ana@email.com",
    habilidades=["Python", "LangChain"],
    anos_exp=3,
)
print(curriculo.nome)        # → "Ana Silva"
print(curriculo.habilidades)  # → ["Python", "LangChain"]
print(curriculo.model_dump())  # → dict Python (Pydantic v2)
print(curriculo.model_dump_json())  # → string JSON
````

### Slide 07 — ValidationError — o Pydantic detecta dados inválidos

````python
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import List, Literal

class Curriculo(BaseModel):
    nome:        str
    anos_exp:    int                           # Deve ser inteiro
    senioridade: Literal["junior","pleno","senior"]  # Enum implícito
    habilidades: List[str]

    # Validator customizado — Pydantic v2
    @field_validator("anos_exp")
    @classmethod
    def anos_positivos(cls, v):
        if v < 0:
            raise ValueError("anos_exp deve ser positivo")
        return v

# Teste de falha — tratar com try/except
try:
    c = Curriculo(
        nome="Carlos",
        anos_exp="dez",         # ❌ str onde int era esperado
        senioridade="ninja",    # ❌ valor fora do Literal
        habilidades=["Python"],
    )
except ValidationError as e:
    print(f"Erros encontrados: {e.error_count()}")
    for erro in e.errors():
        print(f"  Campo: {erro['loc']} | Erro: {erro['msg']}")
    # → Campo: ('anos_exp',) | Erro: Input should be a valid integer
    # → Campo: ('senioridade',) | Erro: Input should be 'junior', 'pleno' or 'senior'
````

### Slide 09 — PydanticOutputParser — schema vira instrução para o modelo

````python
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field
from typing import List

class Curriculo(BaseModel):
    nome:        str       = Field(description="Nome completo")
    email:       str       = Field(description="Endereço de e-mail")
    habilidades: List[str] = Field(description="Habilidades técnicas listadas")
    anos_exp:    int       = Field(description="Anos de experiência")

# 1. Criar o parser com o schema
parser = PydanticOutputParser(pydantic_object=Curriculo)

# 2. O parser gera as instruções de formato automaticamente
print(parser.get_format_instructions())
# → "Return a JSON object with the following schema: {nome: ..., email: ...}"

# 3. Injetar as instruções no prompt com {format_instructions}
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um extrator de dados de currículos.\n{format_instructions}"),
    ("human",  "Extraia as informações deste currículo:\n{curriculo}"),
]).partial(format_instructions=parser.get_format_instructions())

# 4. Chain completa com PydanticOutputParser no final
chain = prompt | llm | parser

# 5. Invocar — retorna objeto Pydantic, não string
resultado = chain.invoke({"curriculo": "João Silva, joao@email.com, 5 anos..."})
print(type(resultado))      # → <class 'Curriculo'>
print(resultado.nome)        # → "João Silva"
print(resultado.anos_exp)    # → 5  (int, não string!)
````

### Slide 11 — format="json" nativo do Ollama — alternativa direta

````python
# Opção 1: format="json" no ChatOllama
# Garante que o modelo retorne JSON válido — mas sem validar o schema
llm_json = ChatOllama(
    model="gpt-oss:120b",
    format="json",   # Ollama garante JSON válido na saída
)

# Com JsonOutputParser — parseia o JSON mas não valida schema
chain_json = prompt | llm_json | JsonOutputParser()
resultado = chain_json.invoke({"curriculo": "texto..."})
print(type(resultado))  # → dict (sem validação de campos)

# Opção 2: format="json" + PydanticOutputParser (combinação recomendada)
llm_json = ChatOllama(model="gpt-oss:120b", format="json")
chain_validado = prompt | llm_json | parser   # parser = PydanticOutputParser
resultado = chain_validado.invoke({"curriculo": "texto..."})
print(type(resultado))  # → <class 'Curriculo'> (objeto Pydantic validado)
````

### Exercício 1 — Modele um schema Pydantic e valide na mão

**Para o professor:** a solução prova a validação antes de plugar o LLM — o caminho bom sai com `model_dump()` tipado e o caminho ruim (valor fora do `Literal`, `anos_exp` como texto) dispara `ValidationError` campo por campo em `e.errors()`. É o mesmo esquema de schema que a chain desta aula passa a exigir na saída.


````python
# Setup da aula — imports e conexão com o Ollama Cloud (Colab Secrets)
!pip install langchain langchain-ollama pydantic -q

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser, JsonOutputParser
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import List, Optional, Literal
import os
from google.colab import userdata
os.environ["OLLAMA_HOST"]    = "https://ollama.com"
os.environ["OLLAMA_API_KEY"] = userdata.get("OLLAMA_API_KEY")

class Curriculo(BaseModel):
    nome:        str       = Field(description="Nome completo")
    email:       str       = Field(description="E-mail")
    habilidades: List[str] = Field(description="Lista de habilidades técnicas")
    anos_exp:    int       = Field(ge=0, le=50, description="Anos de experiência")
    senioridade: Optional[Literal["junior", "pleno", "senior"]] = Field(
        None, description="junior|pleno|senior")

    @field_validator("anos_exp")
    @classmethod
    def anos_positivos(cls, v):
        if v < 0:
            raise ValueError("anos_exp deve ser positivo")
        return v

# Caminho bom — instanciação válida, campos tipados
c = Curriculo(nome="Ana Silva", email="ana@email.com",
              habilidades=["Python", "LangChain"], anos_exp=3, senioridade="pleno")
print(type(c), "→", c.model_dump())

# Caminho ruim — valor fora do Literal e anos_exp como texto: erro campo por campo
try:
    Curriculo(nome="Carlos", email="c@email.com", habilidades=["SQL"],
              anos_exp="dez", senioridade="ninja")
except ValidationError as e:
    print(f"Erros: {e.error_count()}")
    for erro in e.errors():
        print("  ", erro["loc"], "→", erro["msg"])
````

### Exercício 2 — Schema quebrado: análise e correção no código

**Para o professor:** a célula mostra o schema reconstruído — type hints em todos os campos (`str`, `int`, `List[str]`, `Optional[Literal[...]]`), `Field(description=...)` em cada um e `@field_validator` em `tempo_min`. A chain com `format="json"` + `PydanticOutputParser` roda o caminho feliz (`type(r)` → `Receita`, `model_dump()` validado) e o caminho do erro: valor fora do `Literal` dispara `ValidationError` com campo e mensagem em `e.errors()`.


````python
# Setup da aula — imports e conexão com o Ollama Cloud (Colab Secrets)
!pip install langchain langchain-ollama pydantic -q

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser, JsonOutputParser
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import List, Optional, Literal
import os
from google.colab import userdata
os.environ["OLLAMA_HOST"]    = "https://ollama.com"
os.environ["OLLAMA_API_KEY"] = userdata.get("OLLAMA_API_KEY")

class Receita(BaseModel):
    nome:         str       = Field(description="Nome do prato")
    porcoes:      int       = Field(description="Rende quantas porções")
    ingredientes: List[str] = Field(description="Lista de ingredientes")
    dificuldade:  Optional[Literal["facil", "media", "dificil"]] = Field(
        None, description="facil|media|dificil")
    tempo_min:    int       = Field(description="Tempo total em minutos")

    @field_validator("tempo_min")
    @classmethod
    def tempo_positivo(cls, v):
        if v <= 0:
            raise ValueError("tempo_min deve ser positivo")
        return v

parser = PydanticOutputParser(pydantic_object=Receita)

prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um chef. Extraia os dados da receita pedida.\n{format_instructions}"),
    ("human",  "{pergunta}"),
]).partial(format_instructions=parser.get_format_instructions())

llm = ChatOllama(model="gpt-oss:120b", format="json")
chain = prompt | llm | parser

# Caminho feliz — o modelo segue o schema e o Pydantic valida tipos
try:
    r = chain.invoke({"pergunta": "Arroz com frango para 4 pessoas, rápido, ingredientes simples"})
    print(type(r))       # → <class 'Receita'>
    print(r.model_dump())
except ValidationError as e:
    print(f"Erros: {e.error_count()}")
    for erro in e.errors():
        print("  ", erro["loc"], "→", erro["msg"])

# Caminho do erro — valor fora do Literal: ValidationError capturado no except
try:
    Receita(nome="X", porcoes=2, ingredientes=["a"], dificuldade="ninja", tempo_min=10)
except ValidationError as e:
    for erro in e.errors():
        print("  ", erro["loc"], "→", erro["msg"])
        # → ('dificuldade',) | Erro: Input should be 'facil', 'media' or 'dificil'
````

### Exercício 3 — Mesma pergunta, dois parsers

**Para o professor:** duas chains idênticas, exceto pelo parser final — o `JsonOutputParser` devolve `dict` (acesso por `r["campo"]`) e o `PydanticOutputParser` devolve a classe (`r.campo`, com autocompletar e erro imediato em typo). O ponto alto da execução: o input vago ("muitos anos") passa ilesos pelo parser JSON e é barrado na hora pelo Pydantic — a diferença aparece quando o modelo escorrega no schema.


````python
# Setup da aula — imports e conexão com o Ollama Cloud (Colab Secrets)
!pip install langchain langchain-ollama pydantic -q

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser, JsonOutputParser
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import List, Optional, Literal
import os
from google.colab import userdata
os.environ["OLLAMA_HOST"]    = "https://ollama.com"
os.environ["OLLAMA_API_KEY"] = userdata.get("OLLAMA_API_KEY")

class Curriculo(BaseModel):
    nome:        str       = Field(description="Nome completo")
    email:       str       = Field(description="E-mail")
    habilidades: List[str] = Field(description="Habilidades técnicas")
    anos_exp:    int       = Field(description="Anos de experiência")

parser_dict = JsonOutputParser()
parser_pyd  = PydanticOutputParser(pydantic_object=Curriculo)

prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um extrator de currículos.\n{format_instructions}"),
    ("human",  "Extraia deste texto:\n{curriculo}"),
]).partial(format_instructions=parser_pyd.get_format_instructions())

llm = ChatOllama(model="gpt-oss:120b", format="json")

chain_dict = prompt | llm | parser_dict
chain_pyd  = prompt | llm | parser_pyd

texto = "João Silva, joao@email.com, 5 anos de experiência, Python e LangChain"

r1 = chain_dict.invoke({"curriculo": texto})
r2 = chain_pyd.invoke({"curriculo": texto})

print(type(r1))   # → <class 'dict'> — JSON válido, mas SEM validação de schema
print(type(r2))   # → <class 'Curriculo'> — campos tipados e validados

print(r1["nome"], "|", r2.nome)

# E se o modelo escorregar no tipo (anos_exp como "muitos")?
# chain_dict: o dict é entregue como veio — o erro só aparece depois, no uso
# chain_pyd: ValidationError imediata, com campo e mensagem — proteção no fim da chain
try:
    chain_pyd.invoke({"curriculo": "Carlos, carlos@email.com, muitos anos de experiência, Python"})
except Exception as e:
    print(type(e).__name__, "→", e)
````

### Exercício 4 — Schema aninhado com validator

**Para o professor:** a solução modela o schema aninhado completo — `Ponto` com `titulo`/`descricao` dentro de `List[Ponto]`, `Literal` de categoria e `nota_final: float = Field(ge=0, le=10)` — e roda a chain `format="json"` + `PydanticOutputParser` com um input bom, que sai tipado no `model_dump()`. O input ruim (categoria "ninja" e nota 12) dispara `ValidationError` capturado, exibindo em `e.errors()` o campo e a mensagem — validação como proteção real no fim da chain.


````python
# Setup da aula — imports e conexão com o Ollama Cloud (Colab Secrets)
!pip install langchain langchain-ollama pydantic -q

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser, JsonOutputParser
from pydantic import BaseModel, Field, ValidationError, field_validator
from typing import List, Optional, Literal
import os
from google.colab import userdata
os.environ["OLLAMA_HOST"]    = "https://ollama.com"
os.environ["OLLAMA_API_KEY"] = userdata.get("OLLAMA_API_KEY")

class Ponto(BaseModel):
    titulo:    str
    descricao: str

class Avaliacao(BaseModel):
    dominio:       str = Field(description="Domínio avaliado")
    pontos_fortes: List[Ponto] = Field(description="Pontos fortes, cada um com título e descrição")
    pontos_fracos: List[Ponto] = Field(description="Pontos fracos, cada um com título e descrição")
    categoria:     Literal["iniciante", "intermediario", "avancado"] = Field(
        description="Nível do projeto")
    nota_final:    float = Field(ge=0, le=10, description="Nota de 0 a 10")

    @field_validator("nota_final")
    @classmethod
    def nota_valida(cls, v):
        if not (0 <= v <= 10):
            raise ValueError("nota fora do intervalo")
        return v

parser = PydanticOutputParser(pydantic_object=Avaliacao)
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um avaliador de projetos. Responda no schema.\n{format_instructions}"),
    ("human",  "{pergunta}"),
]).partial(format_instructions=parser.get_format_instructions())

chain = prompt | ChatOllama(model="gpt-oss:120b", format="json") | parser

# Input bom → objeto tipado e validado
r = chain.invoke({"pergunta": "Avalie: chatbot de culinária com memória TokenBuffer e saída JSON"})
print(r.model_dump())

# Input ruim (fora do Literal) → ValidationError capturado, campo e mensagem
try:
    Avaliacao(dominio="x", pontos_fortes=[Ponto(titulo="t", descricao="d")],
              pontos_fracos=[], categoria="ninja", nota_final=12)
except ValidationError as e:
    for erro in e.errors():
        print(erro["loc"], "→", erro["msg"])
````
