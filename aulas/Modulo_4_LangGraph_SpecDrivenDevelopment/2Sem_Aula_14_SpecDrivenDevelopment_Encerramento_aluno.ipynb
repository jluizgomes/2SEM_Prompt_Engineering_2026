## 📋 Roteiro do Lab

**Exercício da Aula 14 · 15 minutos · Individual, sem entrega**  
### Escreva sua primeira spec.md
1. Escolha uma ideia — uma melhoria no seu agente do CKP03, uma feature nova ou qualquer problema pequeno que você queira resolver com código.
2. Escreva o Requisito em 2–3 frases: o que o sistema deve fazer, sem mencionar como implementar.
3. Liste 3–4 Critérios de aceite — condições objetivas que provam que a implementação está correta (formato "se X, então Y").
4. Defina o que fica Fora do escopo — o que deliberadamente não será tratado agora, para não inflar a spec.
5. Troque com um colega — ele tenta achar um caso de borda que sua spec não cobre. Essa é a revisão que, num fluxo real, aconteceria contra o código gerado.
